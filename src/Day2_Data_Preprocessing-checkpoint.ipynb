```python
import pandas as pd

df = pd.read_csv("../data/mental_health.csv")
df.head()
```

```python
df.info()
```

```python
df['status'].value_counts()
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df['clean_text'] = df['text'].apply(clean_text)
df[['text', 'clean_text']].head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.drop_duplicates(subset='clean_text', inplace=True)
```

```python
label_mapping = {
    "Normal": 0,
    "Anxiety": 1,
    "Depression": 2,
    "Suicidal": 3
}

df['label'] = df['status'].map(label_mapping)
df[['status', 'label']].head()
```

```python
df['label'].value_counts()
```

```python
df.to_csv("../data/mental_health_clean.csv", index=False)
print("Clean dataset saved successfully!")
```

The dataset used in this project consists of user-generated textual statements related to mental health conditions. Each text sample is labeled into one of four categories: Normal, Anxiety, Depression, and Suicidal.

The dataset is balanced across all classes. Preprocessing steps included text normalization, removal of URLs and special characters, elimination of duplicate records, and encoding of categorical labels into numerical values to prepare the data for deep learning-based classification.
